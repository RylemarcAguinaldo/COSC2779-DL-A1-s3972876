## Training Utilities

Derived from Weeks 3 & 4 Lab notebooks 

```python
def train_one_epoch(model, loader, optimizer, criterion):
    model.train()
    total_loss, correct, total = 0.0, 0, 0
    for x, y in loader:
        x, y = x.to(device), y.to(device)
        outputs = model(x)
        loss = criterion(outputs, y)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * y.size(0)
        _, predicted = torch.max(outputs, 1)
        correct += (predicted == y).sum().item()
        total += y.size(0)
    return total_loss / total, correct / total


def evaluate(model, loader, criterion):
    model.eval()
    total_loss, correct, total = 0.0, 0, 0
    with torch.inference_mode():
        for x, y in loader:
            x, y = x.to(device), y.to(device)
            outputs = model(x)
            loss = criterion(outputs, y)
            total_loss += loss.item() * y.size(0)
            _, predicted = torch.max(outputs, 1)
            correct += (predicted == y).sum().item()
            total += y.size(0)
    return total_loss / total, correct / total


def run_experiment(model, optimizer, criterion, train_loader, val_loader,
                   epochs=10, name='Model'):
    history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': []}
    for epoch in range(epochs):
        train_loss, train_acc = train_one_epoch(model, train_loader, optimizer, criterion)
        val_loss, val_acc = evaluate(model, val_loader, criterion)
        history['train_loss'].append(train_loss)
        history['train_acc'].append(train_acc)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)
        if (epoch + 1) % max(1, epochs // 5) == 0 or epoch == 0:
            print(f'  [{name}] Epoch {epoch+1}/{epochs} — '
                  f'Train: {train_loss:.4f} / {train_acc:.4f} | '
                  f'Val: {val_loss:.4f} / {val_acc:.4f}')
    return history


def plot_experiments(results, show_train=True, show_val=True):
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    for name, hist in results.items():
        epochs = range(1, len(hist['train_loss']) + 1)
        if show_train:
            axes[0].plot(epochs, hist['train_loss'], label=f'{name} (Train)', marker='o', markersize=3)
            axes[1].plot(epochs, hist['train_acc'], label=f'{name} (Train)', marker='o', markersize=3)
        if show_val:
            axes[0].plot(epochs, hist['val_loss'], label=f'{name} (Val)', linestyle='--', marker='x', markersize=3)
            axes[1].plot(epochs, hist['val_acc'], label=f'{name} (Val)', linestyle='--', marker='x', markersize=3)
    for ax, ylabel, title in zip(axes, ['Loss', 'Accuracy'], ['Loss', 'Accuracy']):
        ax.set_xlabel('Epoch')
        ax.set_ylabel(ylabel)
        ax.set_title(title)
        ax.legend(fontsize=8)
        ax.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
```

## Dataset initialisation

### Load Dataset

```python
import os
import re
import zipfile
from PIL import Image
from torch.utils.data import Dataset
from torchvision import transforms

ZIP_PATH = 'A1_data_2026.zip'
DATA_DIR = './data_resized/data_resized'

# ---- Extract zip file ----
if not os.path.isdir(DATA_DIR):
    with zipfile.ZipFile(ZIP_PATH, 'r') as z:
        z.extractall(DATA_DIR)
    print('Extracted to', DATA_DIR)

def load_dataset(root, class_names):
    actual_root = root
    while True:
        subdirectories = [
            d for d in os.listdir(actual_root) 
            if os.path.isdir(os.path.join(actual_root, d)) and not d.startswith('.')
        ]
        if len(subdirectories) == 1 and (subdirectories[0] not in class_names):
            actual_root = os.path.join(actual_root, subdirectories[0])
        else:
            break

    class_to_idx = {cls_name: i for i, cls_name in enumerate(class_names)}
    valid_exts = ('.png', '.jpg', '.jpeg', '.bmp', '.webp')

    all_samples = []
    class_indices = []
    groups = []

    # Example: S13_allClear_HD_instance1_1.jpg
    # Group 1: Subject (S13)
    # Group 2: Gesture Class (allClear)
    # Group 3: Instance ID (instance1)
    pattern = re.compile(r'^(S\d+)_([a-zA-Z0-9]+)_(?:HD_)?(instance\d+)_\d+', re.IGNORECASE)

    for class_name in class_names:
        class_dir = os.path.join(actual_root, class_name)
        if not os.path.isdir(class_dir):
            continue

        target_idx = class_to_idx[class_name]

        for dirpath, _, filenames in os.walk(class_dir):
            for fname in filenames:
                if fname.lower().endswith(valid_exts):
                    img_path = os.path.join(dirpath, fname)
                    
                    match = pattern.match(fname)
                    if match:
                        subject, gesture, instance = match.groups()
                        # Group key combines Subject + Gesture + Instance (e.g., 'S13_alLClear_instance1')
                        group_key = f"{subject}_{gesture}_{instance}"
                    else:
                        # Fallback if filename format varies slightly
                        group_key = os.path.splitext(fname)[0].rsplit('_', 1)[0]

                    all_samples.append((img_path, target_idx))
                    class_indices.append(target_idx)
                    groups.append(group_key)

    return all_samples, class_indices, groups
```

### Get images from dataset

In order to train the model to classify gestures based on instances, the image loader function is designed **to stack five images from the same instance along the channel dimension to form a single image.** This allows for models to process grouped images as a instance rather than individual frames.
https://www.youtube.com/watch?v=A9D6NXBJdwU&t=1s

For some EDA tasks (i.e. visualising instances for each class), a function which only loads individual images will be used instead for simplicity.

```python
class singleImageLoader(Dataset):
    def __init__(self, samples, transform=None):
        """
        samples: List of (image_path, class_index) tuples
        """
        self.samples = samples
        self.transform = transform

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        path, target = self.samples[idx]
        img = Image.open(path).convert('RGB')
        if self.transform is not None:
            img = self.transform(img)
        return img, target


from collections import defaultdict
import torch

class stackedImageLoader(Dataset):
    def __init__(self, samples, transform=None, synchronized=False):
        """
        samples: List of (image_path, class_index, group_key) tuples —
                 one row per FRAME. Frames sharing the same group_key
                 are grouped into a single stacked clip.
        """
        self.transform = transform
        self.synchronized = synchronized

        # Group frame paths by instance (group_key)
        grouped = defaultdict(list)
        labels = {}
        for path, target, group_key in samples:
            grouped[group_key].append(path)
            labels[group_key] = target

        # Sort frames within each instance so temporal order is consistent
        # (assumes filenames sort correctly, e.g. ..._1.jpg ... ..._5.jpg)
        self.instances = []
        for group_key, paths in grouped.items():
            paths_sorted = sorted(paths)
            self.instances.append((paths_sorted, labels[group_key], group_key))

    def __len__(self):
        return len(self.instances)

    def __getitem__(self, idx):
        paths, target, group_key = self.instances[idx]

        # Load all five frames as a list
        frames = [Image.open(p).convert('RGB') for p in paths]

        if self.transform is not None:
            if self.synchronized:
                # Transform expects the whole list, decides randomness once, returns stacked tensor
                clip = self.transform(frames)
            else:
                # Transform expects one image at a time (e.g. deterministic resize/normalize for val)
                frames = [self.transform(f) for f in frames]
                clip = torch.stack(frames)
        else:
            clip = torch.stack([TF.to_tensor(f) for f in frames])
    
        return clip, target
```

### Split dataset

Before data analysis, the dataset needs to be split into train, validation and test sets to avoid any data leakage. The data will be split into 70% training, and 15% each for the validation and test sets. Since there is a small amount of data available with only 1059 instances with some classes having less than 50 instances, allocating as much data as possible for training the dataset is imperative to ensure that the model is able to capture patterns within the data and generalise well against new data. Instances were used as the metric to quantify the amount of available data instead images as the main requirement of the drone model is to recognise gestures based on instances (multiple frames) rather than individual images (single frame).

With the class imbalance in instance distribution, a stratified split is chosen over a random split. This is to preserve the overall class proportion of the dataset. Additionally with the total dataset being relatively small, stratified sampling can also prevent any bias as a random split can result in any of the datasets having more minority class instances compared to the other sets.

The validation method used for this task is cross validation. This method was chosen as having a single hold-out validation split is too unreliable to perform and compare hyperparameter tuning and model comparison. An unlucky seed for either the training or validation set may produce results not representative of the actual performance of the model. For this project, a 5-fold stratified k-fold is performed as with (1 - test_size) / n_splits, where n_splits=0.15 equates to roughly 5 folds.

Final testing will be conducted with hold-out validation.

```python
import numpy as np
from collections import Counter
from sklearn.model_selection import StratifiedGroupKFold


def stratified_group_train_test_split(all_samples, class_indices, groups, test_size=0.15, random_state=42):
    """
    Splits off a held-out test set (instance-grouped, stratified).
    Returns the remaining train+val pool for downstream K-fold CV.
    """
    all_samples = np.array(all_samples, dtype=object)
    class_indices = np.array(class_indices)
    groups = np.array(groups)

    n_splits_1 = max(2, round(1 / test_size))
    sgkf1 = StratifiedGroupKFold(n_splits=n_splits_1, shuffle=True, random_state=random_state)
    trainval_idx, test_idx = next(sgkf1.split(all_samples, class_indices, groups))

    test_samples = [(all_samples[i][0], class_indices[i], groups[i]) for i in test_idx]

    g_trainval, g_test = set(groups[trainval_idx]), set(groups[test_idx])
    assert not (g_trainval & g_test), f"Leakage trainval/test: {g_trainval & g_test}"

    print(f"Trainval: {len(trainval_idx)} images across {len(g_trainval)} instances")
    print(f"Test:     {len(test_samples)} images across {len(g_test)} instances")

    return trainval_idx, test_idx, test_samples


def stratified_group_kfold_splits(all_samples, class_indices, groups, trainval_idx, n_splits=5, random_state=42):
    """
    Performs 5-fold stratified group CV over the trainval pool.
    Yields (fold, train_samples, val_samples, train_idx, val_idx) per fold.
    Indices returned are into the ORIGINAL all_samples/class_indices/groups arrays.
    """
    all_samples = np.array(all_samples, dtype=object)
    class_indices = np.array(class_indices)
    groups = np.array(groups)

    tv_samples = all_samples[trainval_idx]
    tv_classes = class_indices[trainval_idx]
    tv_groups = groups[trainval_idx]

    sgkf = StratifiedGroupKFold(n_splits=n_splits, shuffle=True, random_state=random_state)

    fold_splits = []

    for fold, (train_sub_idx, val_sub_idx) in enumerate(sgkf.split(tv_samples, tv_classes, tv_groups)):
        train_idx = trainval_idx[train_sub_idx]
        val_idx = trainval_idx[val_sub_idx]

        train_samples = [(all_samples[i][0], class_indices[i], groups[i]) for i in train_idx]
        val_samples   = [(all_samples[i][0], class_indices[i], groups[i]) for i in val_idx]

        # Sanity check: no group leakage between this fold's train/val
        g_train, g_val = set(groups[train_idx]), set(groups[val_idx])
        assert not (g_train & g_val), f"Leakage train/val in fold {fold}: {g_train & g_val}"

        print(f"\n=== Fold {fold + 1}/{n_splits} ===")
        print(f"Train: {len(train_samples)} images across {len(g_train)} instances")
        print(f"Val:   {len(val_samples)} images across {len(g_val)} instances")

        fold_splits.append((fold, train_samples, val_samples, train_idx, val_idx))

    return fold_splits
```

Macro F1 score was chosen as the performance metric for this project is due to class imbalance present in the dataset. In skewed datasets, accuracy can be misleading especially if it is only precise in identifying majority tasks.

## Baseline Model

**Justification:** The model architecture for this project is restricted to a 2-D CNN as it to be deployed onto a drone. MobileNetsV2 is known to be computationally efficient unlike ResNet. In addition, the choice for transfer learning with the ImageNets dataset is attributed to the size of the dataset, having only around 727 images after stacking images to form instances.

- Learning rate = 0.001 (1e-3)
- Optimizer = None
- Augmentation = None
- Regularisation = None
- Images normalised to ImageNets mean and std to match network distribution
- Images resized to 224 x 224 to accomodate MobileNetsV2

```python
from itertools import cycle

def plotter(history_hold, metric='loss', ylim=None):
    cycol = cycle('bgrcmk')
    for name, hist in history_hold.items():
        y_train = hist[metric]
        y_val   = hist['val_' + metric]
        x = np.arange(len(y_val))
        c = next(cycol)
        plt.plot(x, y_train, c + '-',  label=name + '_train')
        plt.plot(x, y_val,   c + '--', label=name + '_val')
    plt.legend()
    plt.xlim([1, max(plt.xlim())])
    if ylim is not None:
        plt.ylim(ylim)
    plt.xlabel('Epoch')
    plt.ylabel(metric)
    plt.grid(True)

m_histories = {}
```

```python
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                              roc_auc_score, f1_score)

def run_epoch(model, loader, criterion, optimizer=None, num_classes=13):
    training = optimizer is not None
    model.train(training)
    total_loss, ys, preds, all_probs = 0.0, [], [], []
    ctx = torch.enable_grad() if training else torch.no_grad()
    with ctx:
        for xb, yb in loader:
            xb, yb = xb.to(device), yb.to(device)
            logits = model(xb)
            loss = criterion(logits, yb)
            if training:
                optimizer.zero_grad()
                loss.backward()
                optimizer.step()
            total_loss += loss.item() * len(xb)
            probs = torch.softmax(logits, dim=1).detach().cpu().numpy()
            ys.extend(yb.cpu().numpy())
            preds.extend(logits.argmax(1).detach().cpu().numpy())
            all_probs.extend(probs)
    n = len(ys)
    all_probs = np.array(all_probs)
    metrics = {
        'loss':      total_loss / n,
        'f1_score':  f1_score(ys, preds, average='macro'),
        'accuracy':  accuracy_score(ys, preds),
        'precision': precision_score(ys, preds, average='macro', zero_division=0),
        'recall':    recall_score(ys, preds, average='macro', zero_division=0),
        'auc':       roc_auc_score(ys, all_probs, multi_class='ovr', average='macro')
                     if len(set(ys)) > 1 else float('nan'),
    }
    return metrics


def train_and_track(model, train_loader, val_loader, epochs=10, lr=1e-3,
                    optimizer=None, criterion=None, verbose=True):
    model = model.to(device)
    if optimizer is None:
        optimizer = torch.optim.Adam(
            filter(lambda p: p.requires_grad, model.parameters()), lr=lr)
    if criterion is None:
        criterion = nn.CrossEntropyLoss()  # fallback if no weighted criterion is passed

    hist = {k: [] for k in ['loss', 'f1_score', 'accuracy', 'precision', 'recall', 'auc',
                             'val_loss', 'val_f1_score', 'val_accuracy', 'val_precision', 'val_recall', 'val_auc']}
    for ep in range(epochs):
        tr = run_epoch(model, train_loader, criterion, optimizer=optimizer)
        vl = run_epoch(model, val_loader,   criterion, optimizer=None)
        for k, v in tr.items():        hist[k].append(v)
        for k, v in vl.items():        hist['val_' + k].append(v)
        if verbose:
            print(f'ep {ep+1:2d}/{epochs}  '
                  f'loss {tr["loss"]:.3f}  f1 {tr["f1_score"]:.3f}  acc {tr["accuracy"]:.3f}  |  '
                  f'val_loss {vl["loss"]:.3f}  val_f1 {vl["f1_score"]:.3f}  val_acc {vl["accuracy"]:.3f}  '
                  f'val_prec {vl["precision"]:.3f}  val_rec {vl["recall"]:.3f}  val_auc {vl["auc"]:.3f}')
    return hist
```

### Perform Early Fusion

One of the main requirements for this project is to process an entire instance to classify a gesture instead of individual frames. This is accomplished by performing data fusion, which is the '[combination of] data from multiple sources to produce more accurate, complete, and actionable insights than those derived from individual datasets' (GeeksForGeeks, 2025). Two types of data fusion is early and late fusion.

Early fusion involves the combination of multiple data sources (i.e. multiple images apart of an instance) at the feature level before processed through the model (GeeksForGeeks, 2025). 
Late fusion involves training individual models seperately and combining their predictions at a later stage (GeeksForGeeks, 2025).

For this development pipeline, early fusion is preferred over late fusion. Since the model needs to identify motion-based gestures, the temporal order of the five input frames need to be maintained. Late fusion processes each frame independently and derives an average or most confident prediction out of the frames, removing any temporal information. This means that the model can classify gestures even when gestures are performed out of order or reversed. Early fusion addresses this by sequentially stacking 5 consecutive frames together, changing the image shape from (224,224,3) to (224,224,15); with each frame at a distinct position on the channel. This preserves the temporal information of the instance that the model can train from. 

With the change in image shape, this means that the first convolutional layer has to be modified to accept the image size. This modification also restricts the use of any ImageNets weights for the first layer, requiring the layer to be trained from scratch using the stacked images.

https://cs231n.stanford.edu/2024/slides/2024/lecture_10.pdf

```python
class EarlyFusionMobileNet(nn.Module):
    def __init__(self, num_classes, num_frames=5, freeze_base=True, dropout=0.3):
        super().__init__()
        base = mobilenet_v2(weights=MobileNet_V2_Weights.IMAGENET1K_V2)

        # Replace the first conv layer: 3 input channels -> num_frames * 3
        old_conv = base.features[0][0]  # Conv2d(3, 32, kernel_size=3, stride=2, ...)
        new_conv = nn.Conv2d(
            in_channels=3 * num_frames,
            out_channels=old_conv.out_channels,
            kernel_size=old_conv.kernel_size,
            stride=old_conv.stride,
            padding=old_conv.padding,
            bias=(old_conv.bias is not None),
        )

        # Initialize new conv by REPLICATING pretrained weights across the frame dimension,
        # scaled down — preserves useful ImageNet-learned edge/texture filters instead of
        # starting this layer from random noise.
        with torch.no_grad():
            new_conv.weight[:] = old_conv.weight.repeat(1, num_frames, 1, 1) / num_frames
            if old_conv.bias is not None:
                new_conv.bias[:] = old_conv.bias

        base.features[0][0] = new_conv

        if freeze_base:
            for name, p in base.features.named_parameters():
                if '0.0' not in name:  # keep the new first conv layer trainable
                    p.requires_grad = False

        base.classifier = nn.Sequential(
            nn.Dropout(dropout),
            nn.Linear(1280, num_classes),
        )

        self.backbone = base
        self.num_frames = num_frames

    def forward(self, x):
        # x: (B, T, C, H, W) -> concatenate frames along channel dim
        b, t, c, h, w = x.shape
        x = x.view(b, t * c, h, w)  # (B, 15, H, W)
        return self.backbone(x)
```

```python
from torchvision.models import mobilenet_v2, MobileNet_V2_Weights

def build_tfl_model(freeze_base=True, dropout=0.0, num_classes=13):
    return EarlyFusionMobileNet(num_classes=num_classes, freeze_base=freeze_base,
                                dropout=dropout)
```

```python
def plot_avg_across_folds(m_histories, metric='loss', epochs=10):
    """
    Averages a metric (train and val) across all fold histories in m_histories,
    and plots mean ± std band per epoch.
    """
    train_key = metric
    val_key = 'val_' + metric

    train_curves = np.array([hist[train_key] for hist in m_histories.values()])  # (n_folds, epochs)
    val_curves   = np.array([hist[val_key]   for hist in m_histories.values()])

    train_mean, train_std = train_curves.mean(axis=0), train_curves.std(axis=0)
    val_mean,   val_std   = val_curves.mean(axis=0),   val_curves.std(axis=0)

    x = np.arange(1, epochs + 1)

    plt.figure(figsize=(8, 5))
    plt.plot(x, train_mean, 'b-', label='train (mean)')
    plt.fill_between(x, train_mean - train_std, train_mean + train_std, color='b', alpha=0.15)

    plt.plot(x, val_mean, 'r--', label='val (mean)')
    plt.fill_between(x, val_mean - val_std, val_mean + val_std, color='r', alpha=0.15)

    plt.xlabel('Epoch')
    plt.ylabel(metric)
    plt.title(f'Average {metric} across {len(m_histories)} folds (± std)')
    plt.legend()
    plt.grid(True)
    plt.show()

    print(f"Final epoch — train {metric}: {train_mean[-1]:.4f} ± {train_std[-1]:.4f}  |  "
          f"val {metric}: {val_mean[-1]:.4f} ± {val_std[-1]:.4f}")
```

### Data Augmentation

To improve the generalisation performance of the model, augmentation is performed on the existing images in the train dataset. Augmentation provides additional variability and diversity towards the dataset, which is especially crucial as the amount of instances contained in the dataset is relatively small.

Augmentations of different strengths are experimented on. Light augmentation involves the rotation of images, whereas strong augmentation involves some colour space (i.e. brightness and contrast) and geometric transformations (rotation, zoom). Aggressive augmentations, such as horizontal flipping and extreme rotations will not be performed on the data as many of the gestures are dependent on a position of specific features (i.e. arms up, arms to the left/right, etc).

Since our model is designed to utilise all frames within an instance, we must also perform identical transformations for all frames belonging to the same instance. If each frame within an instance is tranformed individually (random degree of rotation in any direction), then the resulting frame is no longer spatially aligned. This can confuse the model in identifying motion or held poses, making it difficult for the model to recognise any patterns. SynchronizedClipTransform addresses this issue by sampling augmentation per instance, and applying the identical augmentations onto all frames within the instance, preserving spatial information.

```python
# Derived from DL week 6 lab notebook
def plot_experiments(results, show_train=True, show_val=True):
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    for name, hist in results.items():
        epochs = range(1, len(hist['loss']) + 1)
        if show_train:
            axes[0].plot(epochs, hist['loss'], label=f'{name} (Train)', marker='o', markersize=3)
            axes[1].plot(epochs, hist['f1_score'], label=f'{name} (Train)', marker='o', markersize=3)
        if show_val:
            axes[0].plot(epochs, hist['val_loss'], label=f'{name} (Val)', linestyle='--', marker='x', markersize=3)
            axes[1].plot(epochs, hist['val_f1_score'], label=f'{name} (Val)', linestyle='--', marker='x', markersize=3)
    for ax, ylabel in zip(axes, ['Loss', 'F1 Score']):
        ax.set_xlabel('Epoch'); ax.set_ylabel(ylabel); ax.set_title(ylabel)
        ax.legend(fontsize=8); ax.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
```

```python
def plot_single_fold(hist, metric='loss', epochs=10):
    """
    Plots train vs val curves for a single fold (no averaging across folds).
    fold_key: the key identifying which fold's history to plot, e.g. 'fold1'
              or 'Model_tfl_fold1', matching however m_histories is keyed.
    """
    train_key = metric
    val_key = 'val_' + metric

    train_curve = np.array(hist[train_key])
    val_curve   = np.array(hist[val_key])
    x = np.arange(1, epochs + 1)

    plt.figure(figsize=(8, 5))
    plt.plot(x, train_curve, 'b-',  label='train')
    plt.plot(x, val_curve,   'r--', label='val')
    plt.xlabel('Epoch')
    plt.ylabel(metric)
    plt.title(f'{metric} - Single Fold')
    plt.legend()
    plt.grid(True)
    plt.show()

    print(f"Final epoch — train {metric}: {train_curve[-1]:.4f}  |  "
          f"val {metric}: {val_curve[-1]:.4f}")
```

```python
# ---- Configuration & Execution ----
CLASS_NAMES = [
    'AllClear', 'HaveCommand', 'Hover', 'Land', 'LandingDirection', 
    'MoveAhead', 'MoveDownward', 'MoveToLeft', 'MoveToRight', 'MoveUpward', 
    'NotClear', 'SlowDown', 'WaveOff'
]

IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

# Change image size to accomodate MobileNetsV2
IMG_SIZE = 224

# # Augmentation Levels
# aug_configs = {
#     # Light Augmentation - 10 degree rotation
#     'Light Aug': transforms.Compose([
#         transforms.Resize((IMG_SIZE, IMG_SIZE)),
#         transforms.RandomRotation(degrees=10),
#         transforms.ToTensor(),
#         transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
#     ]),
#     # Strong Augmentation - 10 degree rotation, zoom, brightness, contrast
#     'Strong Aug': transforms.Compose([
#         transforms.Resize((IMG_SIZE, IMG_SIZE)),
#         transforms.RandomRotation(degrees=10),
#         # zoom : Scales image between 80% and 120% of its size, then crops back to 224x224
#         transforms.RandomResizedCrop(size=(224, 224), scale=(0.8, 1.2)),
#         # brightness: 0.2 means randomly choosing a factor between [0.8, 1.2]
#         # contrast: 0.2 means randomly choosing a factor between [0.8, 1.2]
#         transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0, hue=0),
#         transforms.ToTensor(),
#         transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
#     ]),
# }

# Synchronise the transforms used for each frame so all images within the frames share identical augmentations
import random
import torchvision.transforms.functional as TF

class SynchronizedClipTransform:
    def __init__(self, img_size, level='none'):
        self.img_size = img_size
        self.level = level  # 'light', 'strong'

    def __call__(self, frames):  # list of 5 PIL images
        angle = random.uniform(-10, 10)
        frames = [TF.rotate(f, angle) for f in frames]

        if self.level == 'strong':
            # Same crop box and color params applied to every frame
            scale = random.uniform(0.8, 1.0)
            i, j, h, w = transforms.RandomResizedCrop.get_params(
                frames[0], scale=(scale, 1.0), ratio=(1.0, 1.0)
            )
            frames = [TF.resized_crop(f, i, j, h, w, (self.img_size, self.img_size)) for f in frames]

            brightness = random.uniform(0.8, 1.2)
            contrast = random.uniform(0.8, 1.2)
            frames = [TF.adjust_brightness(f, brightness) for f in frames]
            frames = [TF.adjust_contrast(f, contrast) for f in frames]

        frames = [TF.resize(f, (self.img_size, self.img_size)) for f in frames]
        frames = [TF.to_tensor(f) for f in frames]
        frames = [TF.normalize(f, IMAGENET_MEAN, IMAGENET_STD) for f in frames]
        return torch.stack(frames)


aug_configs = {
    'Light Aug': SynchronizedClipTransform(IMG_SIZE, level='light'),
    'Strong Aug': SynchronizedClipTransform(IMG_SIZE, level='strong'),
}

val_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])
```

```python
class_indices_arr = np.array(class_indices)
aug_results = {}

# ---- Take only fold 0 ----
fold, train_samples, val_samples, train_idx, val_idx = fold_splits[0]
for aug_name, aug_transform in aug_configs.items():
    print(f"========== Augmentation = {aug_name} ==========")
    train_dataset_aug = stackedImageLoader(train_samples, transform=aug_transform, synchronized=True)
    val_dataset       = stackedImageLoader(val_samples, transform=val_transform)
    
    train_loader = DataLoader(train_dataset_aug, batch_size=32, shuffle=True,  num_workers=0, pin_memory=True)
    val_loader   = DataLoader(val_dataset,       batch_size=64, shuffle=False, num_workers=0, pin_memory=True)
    
    # Fresh model
    model_tfl = build_tfl_model(freeze_base=True, dropout=0.0, num_classes=len(CLASS_NAMES))
    
    # Class weights, computed from fold 0's train labels only
    fold_train_targets = class_indices_arr[train_idx]
    class_weights = compute_class_weight(
        class_weight='balanced',
        classes=np.arange(len(CLASS_NAMES)),
        y=fold_train_targets
    )
    class_weights_tensor = torch.tensor(class_weights, dtype=torch.float32).to(device)
    criterion = nn.CrossEntropyLoss(weight=class_weights_tensor)
    
    # Train
    aug_results[aug_name] = train_and_track(
        model_tfl, train_loader, val_loader,
        epochs=10, lr=1e-3,
        criterion=criterion
    )
```

```python
plot_experiments(aug_results)

# Select the best augmentation and record it
best_aug_name = max(aug_results, key=lambda k: max(aug_results[k]['val_f1_score']))
print(f'Best augmentation: {best_aug_name}')
best_aug_transform = aug_configs[best_aug_name]

# Store in all_results
all_results['v2: +Aug'] = aug_results[best_aug_name]
```
