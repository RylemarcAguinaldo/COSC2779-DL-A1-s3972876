## Load dataset

```python
import os
import zipfile
from PIL import Image
from torch.utils.data import Dataset
from torchvision import transforms

ZIP_PATH = 'A1_data_2026.zip'
DATA_DIR = './data_resized/data_resized'

# ---- Unzip the uploaded file (run once) ----
if not os.path.isdir(DATA_DIR):
    with zipfile.ZipFile(ZIP_PATH, 'r') as z:
        z.extractall(DATA_DIR)
    print('Extracted to', DATA_DIR)


# ---- Custom Dataset with Auto Path-Resolution ----
class CustomFolderDataset(Dataset):
    def __init__(self, root, class_names=None, transform=None):
        self.transform = transform
        self.samples = []

        # Resolve nested wrapper directories automatically
        actual_root = root
        while True:
            subdirectories = [
                d for d in os.listdir(actual_root) 
                if os.path.isdir(os.path.join(actual_root, d)) and not d.startswith('.')
            ]
            # If root contains only 1 folder and it's not a class folder, step inside
            if len(subdirectories) == 1 and (class_names is None or subdirectories[0] not in class_names):
                actual_root = os.path.join(actual_root, subdirectories[0])
            else:
                break

        print(f"Resolved data root: '{actual_root}'")

        if class_names:
            self.classes = class_names
        else:
            self.classes = sorted([d for d in os.listdir(actual_root) if os.path.isdir(os.path.join(actual_root, d))])

        self.class_to_idx = {cls_name: i for i, cls_name in enumerate(self.classes)}
        valid_exts = ('.png', '.jpg', '.jpeg', '.bmp', '.webp')

        # Recursively walk class folders (S[]/Instance[])
        for class_name in self.classes:
            class_dir = os.path.join(actual_root, class_name)
            if not os.path.isdir(class_dir):
                print(f"Warning: Could not find directory for '{class_name}' inside '{actual_root}'")
                continue

            target_idx = self.class_to_idx[class_name]
            for dirpath, _, filenames in os.walk(class_dir):
                for fname in filenames:
                    if fname.lower().endswith(valid_exts):
                        img_path = os.path.join(dirpath, fname)
                        self.samples.append((img_path, target_idx))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        path, target = self.samples[idx]
        img = Image.open(path).convert('RGB')
        
        if self.transform is not None:
            img = self.transform(img)
            
        return img, target


# Basic transforms
base_transform = transforms.Compose([
    transforms.ToTensor(),
])

CLASS_NAMES = [
    'AllClear', 'HaveCommand', 'Hover', 'Land', 'LandingDirection', 
    'MoveAhead', 'MoveDownward', 'MoveToLeft', 'MoveToRight', 'MoveUpward', 
    'NotClear', 'SlowDown', 'WaveOff'
]

# Instantiate dataset
dataset = CustomFolderDataset(
    root=DATA_DIR,
    class_names=CLASS_NAMES,
    transform=base_transform,
)

print(f'Total dataset size: {len(dataset)} images across {len(dataset.classes)} classes')
```
